```python
numbers = [4, 7, 2, 7, 9, 4, 7]

maximum = max(numbers)
count_7 = numbers.count(7)
minimum = min(numbers)
average = sum(numbers) / len(numbers)

print("Maximum:", maximum)
print("Number of 7's:", count_7)
print("Minimum:", minimum)
print("Average:", average)
```

```python
word = "PROGRAMMING"

frequency = {}

for ch in word:
    frequency[ch] = frequency.get(ch, 0) + 1

maximum = max(frequency.values())

print("Character frequencies:", frequency)

print("Most frequent characters:")
for ch in frequency:
    if frequency[ch] == maximum:
        print(ch, ":", maximum)
```

```python
password = input("Enter password: ")

upper = False
lower = False
number = False
special = False

for ch in password:
    if ch.isupper():
        upper = True
    elif ch.islower():
        lower = True
    elif ch.isdigit():
        number = True
    else:
        special = True

if len(password) >= 8 and upper and lower and number and special:
    print("Valid password")
else:
    print("Invalid password")
```

```python
text = "Madam, I'm Adam"

text = text.lower()

clean_text = ""

for ch in text:
    if ch.isalnum():
        clean_text += ch

reverse_text = clean_text[::-1]

print("Original:", text)
print("After removing spaces/punctuation:", clean_text)
print("Reverse:", reverse_text)

if clean_text == reverse_text:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
scores = [72, 45, 91, 63, 28, 84, 56]

scores.sort()

print("Ascending order:", scores)
```

```python
scores = [72, 45, 91, 63, 28, 84, 56]

n = len(scores)

for i in range(n):
    for j in range(n - i - 1):
        if scores[j] > scores[j + 1]:
            scores[j], scores[j + 1] = scores[j + 1], scores[j]

print("Ascending order:", scores)
```

```python
scores = [72, 45, 91, 63, 28, 84, 56]

n = len(scores)

for i in range(n):
    min_index = i

    for j in range(i + 1, n):
        if scores[j] < scores[min_index]:
            min_index = j

    scores[i], scores[min_index] = scores[min_index], scores[i]

print("Ascending order:", scores)
```
