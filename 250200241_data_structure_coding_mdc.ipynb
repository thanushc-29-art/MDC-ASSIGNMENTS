```python
numbers = []

for i in range(2000, 3201):
    if i % 7 == 0 and i % 3 != 0:
        numbers.append(str(i))

print(",".join(numbers))
```

```python
n = int(input("Enter a number: "))

result = {}

for i in range(1, n + 1):
    result[i] = i * i

print(result)
```
